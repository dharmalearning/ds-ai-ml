    * Iterating over a range

```python
## for loop
range(5)

for i in range(5):
    print(i)
```

```python
for i in range(10,1,-2):
    print(i)
```

```python
## iterating over a string

str = "Dharma"
for i in str:
    print(i)
```

```python
## while loop

i = 0
while i < 5:
    print(i)
    i += 1
```

```python
## break

for i in range(10):
    if i == 5:
        break
    print(i)

    
```

```python
## continue
for i in range(5):
    if i == 2:
        continue
    print(i)
```

```python
## Pass - pass and continue are similar in that they both skip the current iteration of a loop. 
## The difference is that pass does not skip the current iteration, it just does nothing.

for i in range(5):
    if i == 2:
        pass
    print(i)
```
