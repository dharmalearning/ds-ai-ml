```python
## if statement
age = 35
if age > 20:
    print("You are older than 20")
    print("You are an adult")
    if(age > 30):
        print("You are older than 30")
```

```python
## else statement
age = 20

if age > 18:
    print("You are older than 18")
else:
    print("You are younger than 18")
```

```python
## if elif else statement

age = 20
if age < 13:
    print("You are a child")
elif age < 18:
    print("You are a teenager")
else:
    print("You are an adult")
```

```python
## Nested if statement

## if statement
age = 35
if age > 20:
    print("You are older than 20")
    if(age == 30):
        print("You are  30")
    else:
        print("You are not 30")
```
