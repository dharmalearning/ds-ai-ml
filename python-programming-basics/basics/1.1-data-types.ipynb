# Data Types

1. Definition
    * Data types are a classification of data which tell the compiler or interpreter how the programmer intends to use the data.
    * They determine the type of operations that can be performed on the data, the values that the data can take and the amount of memory needed to store the data.

2. Importance of Data Types in programming

    Explanation:

    * Data types ensure that data is stored in an efficient way
    * They help in performing correct operation on data
    * Proper use of data types can prevent errors and bugs in the program.

```python
## Integer example
age=35
type(age)
```

```python
## Floating point datatype
height=5.9
print(height)
type(height)
```

```python
## String datatype
name="John Doe"
print(name)
type(name)
```

```python
## boolean datatype
is_student=True
print(is_student)
type(is_student)
```

```python
a=10
b=20
print(a == b)
type(a == b)
```

# Operators in Python

1. Arithmetic Operator
    * Addition
    * Substraction
    * Multiplication
    * Division
    * Floor Division
    * Modulus
    * Exponentiation
2. Comparison Operators
    * Equal to 
    * Not Equal to
    * Greater than
    * Less than
    * Greater than or Equal To
    * Less than or Equal To
3. Logical Operators
    * AND
    * OR
    * NOT

```python
## Arithmetic operators

a = 10
b = 5

print("addition",a + b) # addition
print("subtraction",a - b) # subtraction
print("multiplication",a * b) # multiplication
print("division",a / b) # division
print("modulus",a % b) # modulus
print("exponentiation",a ** b) # exponentiation
print("floor division",a // b) # floor division
```

```python
## Comparison operators
a = 10
b = 5
print(a == b) # equal to
print(a != b) # not equal to
print(a > b) # greater than
print(a < b) # less than
print(a >= b) # greater than or equal to
print(a <= b) # less than or equal to
```

```python
## Logical operators
a = True
b = False
print(a and b) # logical AND
print(a or b) # logical OR
print(not a) # logical NOT
```

```python
## Bitwise operators
a = 10 # 1010 in binary
b = 4 # 0100 in binary
print(a & b) # bitwise AND
print(a | b) # bitwise OR
print(a ^ b) # bitwise XOR
print(~a) # bitwise NOT
print(a << 1) # left shift
print(a >> 1) # right shift
```

```python
## Identity operators
a = [1, 2, 3]
b = a
c = a.copy()
print(a is b) # True, because a and b refer to the same object
print(a is c) # False, because a and c refer to different objects
print(a is not c) # True, because a and c refer to different objects
```

```python
## Membership operators
a = [1, 2, 3, 4, 5]
print(1 in a) # True, because 1 is in the list
print(6 not in a) # True, because 6 is not in the list
print(1 not in a) # False, because 1 is in the list
print(6 in a) # False, because 6 is not in the list
```
