# Importing Modules in Python: Modules and Packages

In Python, modules and packages help prganize and reuse code. Here's a comprehensive guide on how to import them.

```python
import math
print(math.sqrt(4))
```

```python
from math import sqrt, pi
print(sqrt(4))
print(pi)
```

```python
## using alias with import
import numpy as np
```

```python
## importing from self created package

from package.maths import addition
print(addition(2, 3))

from package import maths
print(maths.addition(2, 3))
```
