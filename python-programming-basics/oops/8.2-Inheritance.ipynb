# Inheritance in Python

Inheritance is a fundamental concept in Obkect oriented programming that allows a class to inherit attributes and methods from another class.

```python
## Inheritance
## Parent class
class Car:
    def __init__(self, windows,doors,enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype

    def drive(self):
        print(f"The car is driving : {self.enginetype} engine")

## Child class
class Tesla(Car):
    def __init__(self, windows, doors, enginetype, autopilot):
        super().__init__(windows, doors, enginetype)
        self.autopilot = autopilot
    
    def drive(self):
        print(f"The Tesla is driving : {self.enginetype} engine")
        print(f"Autopilot status: {self.autopilot}")

car2 = Tesla(4, 4, "Electric", True)
car2.drive()
```

```python
## Multiple Inheritance
## When a class is derived from more than one base class

# Base class 1
class Animal:
    def __init__(self,name):
        self.name = name
    
    def speak(self):
        print(f"{self.name} makes a sound")

# Base class 2
class Pet:
    def __init__(self, owner):
        self.owner = owner
    
    def show_owner(self):
        print(f"This pet belongs to {self.owner}")

## Derived class

class Dog(Animal, Pet):
    def __init__(self,name,owner):
        Animal.__init__(self,name)
        Pet.__init__(self,owner)
    
    def speak(self):
        print(f"{self.name} barks")

    def show_owner(self):
        print(f"{self.name} belongs to {self.owner}")


toto = Dog("Toto", "Alice")
toto.speak()
toto.show_owner()
```
