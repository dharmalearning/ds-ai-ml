# Encapsulation And Abstraction

Encapsulation and abstraction are two fundamental principles of Object oriented Programming that help in designing robust, maintainable and reusable code. Encapsulation involves bundlining data and methods that operate on the data within a single unit, while abstraction involves hiding complex implementation details and exposing only the necessary features.

## Encapsulation

Encapsulation is the concept of wrapping data (Variables) and methods (functions) together as a single unit. It resticts direct access to some of the object's components, which is a means of preventing accidential interference and misuses of the data.

```python
## Encapsulation with Getters and Setters
## Public, protected, and private attributes

class Person:
    def __init__(self, name, age):
        self.name = name # Public attribute
        self.age = age

person = Person("Alice", 30)
print(person.name)  # Accessing public attribute
```

```python
dir(person)  # List all attributes and methods of the object
```

```python
## Private variables use __ double underscore befor the attribute name
class Person1:
    def __init__(self, name, age):
        self.__name = name  # Private attribute
        self.__age = age

    def get_name(self):
        return self.__name

dharma = Person1("Dharma", 30)

print(dharma.get_name())  # Accessing private attribute using getter
```

```python
## Protected variable use _ single underscore before the attribute name
class Person:
    def __init__(self, name, age):
        self._name = name
        self._age = age
    
class Employee(Person):
    def __init__(self,name, age, salary):
        super().__init__(name, age)
        self._salary = salary

emp = Employee("John", 30, 50000)
print(emp._name)  # Accessing protected attribute
```
