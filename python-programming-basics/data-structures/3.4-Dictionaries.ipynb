## Introduction to Dictionaries

Dictionaries are unorderd collections of items. They store data in key-value pairs. Key must be unique and immutable (e.g strings, numbers or tuples) while values can be of any type.

## Creating Dictionaries

```python
## Creating dictionaries

empty_dict = {}
print(type(empty_dict))  # <class 'dict'>
```

```python
empty_dict = dict()
print(type(empty_dict))  # <class 'dict'>
```

```python
student = { "name": "Dharumaraj", "age": 22, "course": "Python" }
print(student)
```

## Accessing Dictionaries

```python
## Accessing dictionary values
print(student["name"])
print(student["age"])
```

```python
## Accessing using get method
print(student.get("name"))

## if the key value pair not available in dictionaries, when using get it will return None
print(student.get("address"))  # None
print(student.get("address", "Not Available"))  # Not Available is default value
```

```python
## update the value of a key

student["name"] = "Dharu"
print(student)  # {'name': 'Dharu', 'age': 22, 'course': 'Python'}
```

```python
## deleting a key value pair
del student["name"]
print(student)  # {'age': 22, 'course': 'Python'}
```

```python
#Dictionary Methods
# 1. keys() - returns a list of all the keys in the dictionary
print(student.keys())  # dict_keys(['age', 'course'])

# 2. values() - returns a list of all the values in the dictionary
print(student.values())  # dict_values([22, 'Python'])

# 3. items() - returns a list of tuples, each tuple containing a key-value pair
print(student.items())  # dict_items([('age', 22), ('course', 'Python')])

# 4. pop() - removes the specified key and returns its value # 22
print(student.pop("age"))  # 22
print(student)  # {'course': 'Python'}
```

```python
## Shallow copy
student_copy = student
print(student_copy)  # {'course': 'Python'}
print(student)  # {'course': 'Python'}

student_copy["course"] = "Java"
print(student_copy)  # {'course': 'Java'}
print(student)  # {'course': 'Java'}

## shallow copy using copy() method
student = { "name": "Dharumaraj", "age": 22, "course": "Python" }
student_copy = student.copy()
print(student_copy)  # {'name': 'Dharumaraj', 'age': 22, 'course': 'Python'}
print(student)  # {'name': 'Dharumaraj', 'age': 22, 'course': 'Python'}
student_copy["course"] = "Java"
print(student_copy)  # {'name': 'Dharumaraj', 'age': 22, 'course': 'Java'}
print(student)  # {'name': 'Dharumaraj', 'age': 22, 'course': 'Python'}
```

```python
## Iterating through a dictionary

## ITerating over keys

for keys in student.keys():
    print(keys)  # name age course

for values in student.values():
    print(values)  # Dharumaraj 22 Python

for key, value in student.items():
    print(key, value)  # name Dharumaraj age 22 course Python
```

```python
students = {
    "student1": { "name": "Dharumaraj", "age": 22, "course": "Python" },
    "student2": { "name": "John", "age": 23, "course": "Java" },
    "student3": { "name": "Smith", "age": 24, "course": "C++" }
}

student_copy = students.copy()
print(student_copy)  # {'student1': {'name': 'Dharumaraj', 'age': 22, 'course': 'Python'}, 'student2': {'name': 'John', 'age': 23, 'course': 'Java'}, 'student3': {'name': 'Smith', 'age': 24, 'course': 'C++'}}
```

```python
students["student1"]["name"] = "Dharu"
print(students)  # {'student1': {'name': 'Dharu', 'age': 22, 'course': 'Python'}, 'student2': {'name': 'John', 'age': 23, 'course': 'Java'}, 'student3': {'name': 'Smith', 'age': 24, 'course': 'C++'}}
print(student_copy)  # {'student1': {'name': 'Dharumaraj', 'age': 22, 'course': 'Python'}, 'student2': {'name': 'John', 'age': 23, 'course': 'Java'}, 'student3': {'name': 'Smith', 'age': 24, 'course': 'C++'}}
```

```python
## Deep copy using copy module
import copy
student_copy = copy.deepcopy(students)
students["student1"]["name"] = "Maru"
print(students)  # {'student1': {'name': 'Dharu', 'age': 22, 'course': 'Python'}, 'student2': {'name': 'John', 'age': 23, 'course': 'Java'}, 'student3': {'name': 'Smith', 'age': 24, 'course': 'C++'}}
print(student_copy)  # {'student1': {'name': 'Dharumaraj', 'age': 22, 'course': 'Python'}, 'student2': {'name': 'John', 'age': 23, 'course': 'Java'}, 'student3': {'name': 'Smith', 'age': 24, 'course': 'C++'}}
```

```python
for student, details in students.items():
    print(f"{student}")
    for key,value in details.items():
        print(f"{key}: {value}")
```

```python
## Dictionary Comprehension
squares = {x: x**2 for x in range(1, 11) if x % 2 == 0}
print(squares)  # {1: 1, 2: 4, 3: 9, 4: 16, 5: 25, 6: 36, 7: 49, 8: 64, 9: 81, 10: 100}
```

```python
## Practical examples

numbers = [1, 2, 3, 4, 5, 1, 1, 2,2,2,3,4,5]
frequency = dict()

for number in numbers:
    if number in frequency:
        frequency[number] += 1
    else:
        frequency[number] = 1

print(frequency)  # {1: 4, 2: 5, 3: 3, 4: 2, 5: 2}

```

```python
## Merging dictionaries
dict1 = {"a": 1, "b": 2}
dict2={"b":3, "c": 5}
merged_dict = {**dict1, **dict2}

print(merged_dict)  # {'a': 1, 'b': 2, 'c': 4}
```

```python
## words frequency in a string
string = "hello world hello"
words = string.split()
words_count = dict()

for word in words:
    words_count[word] = words_count.get(word, 0) + 1

print(words_count)  # {'hello': 2, 'world': 1}
```

```python
## check tuple is a palindrome
def is_palindrome(tup):
    return tup == tup[::-1]
print(is_palindrome((1, 2, 3, 2, 1)))  # True
```

```python
## merge dictionaries in list withe overlapping keys
dict1 = [{'a': 1, 'b': 2}, {'b': 3, 'c': 4}, {'c': 5, 'd': 6}]
dict2 = dict()

for d in dict1:
    for key,value in d.items():
        if key in dict2:
            dict2[key] += value
        else:
            dict2[key] = value
print(dict2)  # {'a': 1, 'b': 3, 'c': 4} 
```
