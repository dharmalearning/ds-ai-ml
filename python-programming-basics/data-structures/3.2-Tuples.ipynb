## Inroduction to Tuples

Tuples are ordered collection of items that are immutable. They are similar to lists, but their immutability makes them different.

```python
## Creating tuple

empty_tuple = ()
print(type(empty_tuple))  # <class 'tuple'>
```

```python
empty_tuple = tuple([1,2,3,4,5])
print(type(empty_tuple))  # <class 'tuple'>
```

```python
## mixed tuples
mixed_tuple = (1, 2.0, '3', [4, 5], (6, 7))
print(mixed_tuple)  # (1, 2.0, '3', [4, 5], (6, 7))
```

```python
## Accerssing tuple elements
print(mixed_tuple[0])  # 1
print(mixed_tuple[1])  # 2.0
print(mixed_tuple[:2])  # '3'
```

```python
concatenated_tuple = mixed_tuple + (1,8, 9)
print(concatenated_tuple)  # (1, 2.0, '3', [4, 5], (6, 7), 8, 9)
```

```python
## Tuple methods
print(concatenated_tuple.count(1))  # 1
print(concatenated_tuple.index(2.0))  # 1
```
