```python
## Accessing List ITems

fruits = ["apple", "banana", "cherry"]
print(fruits[0])  # apple
print(fruits[-1])  # -1 refers to the last element
print(fruits[1:2])  # banana, cherry 1 inclusinve and 3 exclusive
print(fruits[1:])  # banana, cherry
print(fruits[-3:])  # cherry, banana
```

```python
## Appending to list

fruits.append("orange")
print(fruits)  # ['apple', 'banana', 'cherry', 'orange']
```

```python
fruits.insert(1, "kiwi")
print(fruits)  # ['apple', 'kiwi', 'banana', 'cherry', 'orange']
```

```python
fruits.remove("banana")
fruits
```

```python
## Remove and return the last element
popped_fruit = fruits.pop()
print(popped_fruit)  # orange
```

```python
## Get index of any fruits by element
print(fruits.index("kiwi"))  # 1
```

```python
## Get the count of a specific fruit
print(fruits.count("kiwi"))  # 1
```

```python
## Sorting

fruits.sort()
print(fruits)  # ['apple', 'cherry', 'kiwi']
```

```python
## Reersing the elements in the list
fruits.reverse()
print(fruits)
```

```python
## Remving all elements from the list
fruits.clear()
print(fruits)  # []
```

## Slicing List

```python
## Slicing lists
numbers = [1, 2, 3, 4, 5, 7, 8, 9, 10]
print(numbers[2:5])
print(numbers[:5])
print(numbers[5:])
print(numbers[::2])  ## :: means all the elements and ::2 mmeans 2 refers to the jump count, in this case, the output will be 1,3,5,8,10
print(numbers[::-1]) ## returns the list in reverse order
```

```python
## Iterating with index
for index,number in enumerate(numbers):
    print(index, number)
```

## List Comprehension

```python
## List Comprehension
lst = []
for i in range(10):
    lst.append(i**2)
print(lst)  # [0, 1, 4, 9, 16, 25, 36, 49, 64, 81]


## With pythons list comprehension
lst = [i**2 for i in range(10)]
print(lst)
```

##### List Comprehension
Basic Syntax             [expression for item in iterable]


With Conditional logic   [expression for item in iterable if condition]

Nested List Comprehension [expression for item in iterable for item in iterable]

```python
## Basic list comprehension

square = [num**2 for num in range(10)]
print(square)
```

```python
## List comprehension with condition
square = [num**2 for num in range(10) if num % 2  == 0 and num != 4]
print(square)  # [0, 4, 16, 36, 64]
```

```python
lst = [1, 2, 3, 4, 5]
lst1 = ['a','b','c']

pairs = [(x,y) for x in lst for y in lst1]
print(pairs)
```
